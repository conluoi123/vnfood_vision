# Tổng kết quá trình của ResNet50

```
import pandas as pd
import matplotlib.pyplot as plt
import os

# Đường dẫn tới file log trên Colab
log_path = "/content/vnfood_vision/checkpoints/resnet50/training_log.csv"

if not os.path.exists(log_path):
    print("Chưa tìm thấy file log. Hãy đợi train xong vài Epoch rồi chạy lại nhé!")
else:
    # Đọc file CSV
    df = pd.read_csv(log_path)

    # Thiết lập khung vẽ (2 biểu đồ cạnh nhau)
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

    # 1. Biểu đồ Loss (Độ trễ/Lỗi)
    ax1.plot(df['epoch'], df['train_loss'], label='Train Loss', marker='o', color='blue')
    ax1.plot(df['epoch'], df['val_loss'], label='Validation Loss', marker='o', color='red')
    ax1.set_title('Đồ thị Loss qua các Epoch')
    ax1.set_xlabel('Epoch')
    ax1.set_ylabel('Loss')
    ax1.legend()
    ax1.grid(True, linestyle='--', alpha=0.6)

    # 2. Biểu đồ Accuracy (Độ chính xác)
    ax2.plot(df['epoch'], df['train_acc'], label='Train Accuracy', marker='o', color='blue')
    ax2.plot(df['epoch'], df['val_acc1'], label='Validation Top-1 Acc', marker='o', color='red')
    ax2.set_title('Đồ thị Accuracy qua các Epoch')
    ax2.set_xlabel('Epoch')
    ax2.set_ylabel('Accuracy (%)')
    ax2.legend()
    ax2.grid(True, linestyle='--', alpha=0.6)

    plt.tight_layout()
    plt.show()
```
